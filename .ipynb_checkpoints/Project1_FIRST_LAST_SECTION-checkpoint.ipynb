### Introduction to Vector Embeddings

The intersection of linear algebra and AI represents one of the most critical knowledge domains for modern applications. Over the past month, we have explored the fundamental properties of vector spaces, the mechanics of matrix-vector multiplication, and the power of solving linear systems. Project 1 will bridge the gap between what we have been mastering in class with the realities of Large Language Models and recommendation systems.

So far, we have often viewed vectors as coordinates and use them when solving systems like $Ax = b$. While foundational, this approach understates the representational power of an ordered list of numbers to encode the meaning of words or ideas. Embedding layers in LLMs, protein structure predictions, and recommendation systems all have vectors of the primary data structure and dot products as the primary operation.

### The Problem

When a streaming service like Spotify or Apple Music recommends a new track to you, it isn't guessing. It is using your listening history to navigate a massive, multi-dimensional database. But how does an algorithm compare a Taylor Swift song to Bad Bunny? How do we calculate the "distance" between an indie singer-songwriter track and a classic country song?

The answer lies in vector embedding. By representing every song as a vector in a high-dimensional vector space, we can use the geometric tools of linear algebra like dot products, projections, and change of basis to mathematically define similarity, style, and genre.

## Normalization

To make Song A and Song B comparable from our example above, we need to remove the influence of magnitude. We do this by normalizing every vector in our dataset to be
$$ \mathbf{u} = \frac{\mathbf{v}}{\| \mathbf{v} \|} $$

Every song vector is shrunk or stretched until it has a length of exactly 1. By forcing all songs to live on the same "surface", distance is now purely a measure of the angle between styles.

Before normalization:
1. $\textbf{Song A (Loud):}$ $\mathbf{v}_A = \begin{bmatrix} 10 \\ 4 \end{bmatrix}$ (High Volume, Moderate Tempo)
2. $\textbf{Song B (Quiet):}$ $\mathbf{v}_B = \begin{bmatrix} 5 \\ 2 \end{bmatrix}$ (Med Volume, Moderate Tempo)

After normalization, the vectors are identical:

$$\mathbf{v}_{A/B} = \begin{bmatrix} 0.928 \\ 0.371 \end{bmatrix}$$

*Note that we first standardized the song vectors to get each of the nine features on the same scale. We will not cover this in any detail.

## Ready for Dot Products

Now that every song is a unit vector, we have unlocked a special property of the dot product. Section 1.2 of Strang defines the geometric definition of the dot product as

$$
\mathbf{v}\cdot\mathbf{w}=\lVert \mathbf{v} \rVert\,\lVert \mathbf{w} \rVert \cos\theta,
$$

which reveals the utility of the operation. If vectors are normalized such that $\lVert \mathbf{v} \rVert = 1$, then the dot product simplifies to $\cos\theta$. This means the dot product of two song vectors gives us the Cosine Similarity directly.

### Feature Isolation

So far, we have relied on the nine features provided in the Spotify dataset. We can easily answer which song is the most acoustic by looking at that component in our vector. What if we want to ask more complex questions like "what are the vocals like is this song?" There is no single column for vocals. To do this, we construct a new axis in our 9D space which points towards the abstract concept of "vocals".

In Section 4.2, Strang will introduce the projection matrix

$$
p=\frac{aa^{\mathsf T}}{a^{\mathsf T}a}
$$

which projects a vector onto the line spanned by $a$. In the context of music search, this allows for the isolation of specific components of the song. The 'vocal axis' in the vector space we will create is represented by
the vector $\mathbf{v}_{\text{vocal}}$.

Imagine you have a song vector $\mathbf{b}$ that contains both vocal and instrumental components. The vocal component can be isolated by projecting $\mathbf{b}$ onto the vocal axis. We want to break the vector $\mathbf{b}$ into two perpendicular components:
$$ \mathbf{b} = \mathbf{p} + \mathbf{e} $$

1. $\mathbf{p}$ is the Vocal Component (the part of the song parallel to our vocal archetype).
2. $\mathbf{e}$ is the Instrumental Component (the part of the song orthogonal to the vocals).

### Calculating the Projection

The projection of a vector $\mathbf{b}$ onto a unit vector $\mathbf{v}$ is the "shadow" that $\mathbf{b}$ casts along that direction. Since $\mathbf{v}_{vocal}$ is a unit vector, the formula for the projection $\mathbf{p}$ is:

$$ \mathbf{p} = \text{proj}_{\mathbf{v}_{vocal}}\mathbf{b} = (\mathbf{b} \cdot \mathbf{v}_{vocal}) \mathbf{v}_{vocal} $$

Recall from previously that the dot product $(\mathbf{b} \cdot \mathbf{v}_{vocal})$ is a scalar that tells us the similarity between the song and the vocal archetype. If the dot product is high, the scalar is large, and the shadow $\mathbf{p}$ is long meaning the song is very vocal. If the dot product is zero, the scalar is zero, and $\mathbf{p} = \mathbf{0}$.

Once we have the vocal part $\mathbf{p}$, finding the instrumental part is a simple matter of vector subtraction. This is the error vector:
$$ \mathbf{e} = \mathbf{b} - \mathbf{p} $$

Geometrically, we have subtracted the "vocal-ness" out of the song.

Note: In class, we learned that the null space of a transformation is the set of all vectors that map to zero. Any song that is already purely instrumental will have a projection $\mathbf{p} = \mathbf{0}$.  Therefore, purely instrumental songs live in the null space of the projection onto $\mathbf{v}_{vocal}$.

### Visualization and Dimensionality Reduction

Finally, now that we have embedded our songs and artists in $\mathbb{R}^{9}$ how can we visualize our embeddings of songs? Humans are limited to three spatial dimensions, yet AI embeddings often operate in hundreds or thousands. Principal Component Analysis (PCA) is the primary tool used by AI engineers to 'squash' these high-dimensional spaces into a form that can be visually inspected while retaining the most important patterns in the data. We will learn about PCA in future sections of the course. Run the code below for your favorite set of artists (minimum 10) and interpret the results.

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

artists_10 =

mask = df['track_artist'].isin(artists_10).to_numpy()
labels = df.loc[mask, 'track_artist'].to_numpy()

def plot_cloud_and_centroids(Z, title):
    plt.figure(figsize=(10,7))
    plt.scatter(Z[:, 0], Z[:, 1], s=10, alpha=0.10)

    # centroids
    for artist in artists_10:
        pts = Z[labels == artist]
        if len(pts) == 0:
            continue
        c = pts.mean(axis=0)
        plt.scatter(c[0], c[1], s=250, edgecolors='black')
        plt.text(c[0], c[1], " " + artist, fontsize=9)

    plt.title(title)
    plt.grid(alpha=0.2)

    # Fixed axis limits
    plt.xlim(-4, 3)
    plt.ylim(-3, 3)

    plt.tight_layout()
    plt.xlabel("Component 1")
    plt.ylabel("Component 2")
    plt.show()

# PCA on standardized features only
Z_scaled = PCA(n_components=2, random_state=0).fit_transform(A[mask])

plot_cloud_and_centroids(Z_scaled, "Principal Component Analysis")
```
