### Introduction to Vector Embeddings

The intersection of linear algebra and AI represents one of the most critical knowledge domains for modern applications. Over the past month, we have explored the fundamental properties of vector spaces, the mechanics of matrix-vector multiplication, and the power of solving linear systems. Project 1 will bridge the gap between what we have been mastering in class with the realities of Large Language Models and recommendation systems.

So far, we have often viewed vectors as coordinates and use them when solving systems like $Ax = b$. While foundational, this approach understates the representational power of an ordered list of numbers to encode the meaning of words or ideas. Embedding layers in LLMs, protein structure predictions, and recommendation systems all have vectors of the primary data structure and dot products as the primary operation.

### The Problem

When a streaming service like Spotify or Apple Music recommends a new track to you, it isn't guessing. It is using your listening history to navigate a massive, multi-dimensional database. But how does an algorithm compare a Taylor Swift song to Bad Bunny? How do we calculate the "distance" between an indie singer-songwriter track and a classic country song?

The answer lies in vector embedding. By representing every song as a vector in a high-dimensional vector space, we can use the geometric tools of linear algebra like dot products, projections, and change of basis to mathematically define similarity, style, and genre.


### Project Goal

This notebook will guide you through the following tasks:



1. $\textbf{Construct a Song Space}$: Build a 9-dimensional vector space ($\mathbb{R}^9$) where every track is a unique vector defined by nine musical features.
2. $\textbf{Normalize}$: Apply transformations to ensure that these features are comparable.
3. $\textbf{Measure Similarity}$: Use the dot product to find songs that share a "stylistic direction", proving that similarity is a matter of angles, not just numbers.
4. $\textbf{Engineer Semantic Directions}$: Use orthogonal projections to separate the vocal and instrumental components of a song.
5. $\textbf{Visualize Song Embeddings}$: Use principal component analysis to visualize our 9D song vector in just 2D.






### Dataset

We can translate any complex entity like a song, a system of equations, an image, or a clinical trial result into a specific list of numbers. To apply key ideas from Strang to real world data, we will use a dataset of Spotify tracks over a range of 125 different genres.
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from IPython.display import display
from sklearn.preprocessing import StandardScaler

# Load the Spotify Master Dataset
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-01-21/spotify_songs.csv"
df_spotify = pd.read_csv(url)

feature_names = ['danceability', 'energy', 'loudness', 'speechiness',
                 'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo']

df = df_spotify[['track_name', 'track_artist'] + feature_names].dropna().reset_index(drop=True)

nice = (
    df.head(15)
    .style
    .set_caption("df_spotify (first 15 rows)")
    .format(precision=3)
    .set_properties(**{
        "text-align": "left",
        "white-space": "nowrap",
        "padding": "6px 10px",
        "border": "1px solid #ddd",
    })
    .set_table_styles([
        {"selector": "caption", "props": [("caption-side", "top"), ("font-weight", "600"), ("font-size", "14px")]},
        {"selector": "th", "props": [("background-color", "#f5f5f5"), ("border", "1px solid #ddd")]},
    ])
)
display(nice)

A = df[feature_names].to_numpy()
A = StandardScaler().fit_transform(A)
m, n = A.shape
print(f"Processed {m} tracks with {n} latent features.")
```

### Vectors as Feature Representations

The concept of a linear combination $cv + dw$ in this music context is the mechanism of semantic mixing. If we take the vector for "Fleetwood Mac" and add the vector for "Electronic Distortion," we transform the point in the vector space. The new location might be close to the vector for "Calvin Harris."

However, we have a geometric issue that prevents us from comparing songs effectively right away.

Imagine two rock songs.

1. $\textbf{Song A}$ is a studio recording of an 80s rock track. It is loud, fast, and energetic. Its vector coordinates are large positive numbers.
2. $\textbf{Song B}$ is the "Spotify Studio Unplugged" acoustic version of the exact same track. It has the same tempo and the same musical structure, but it is quieter and less intense. Its coordinates are smaller numbers.

If we measure the Euclidean distance between these two vectors, they will appear far apart because their magnitudes are different. However, musically, they are the same song. In the song vector space, the magnitude of a vector represents its intensity (how loud/extreme it is), while the direction of the vector represents its style (the ratio of features). To find similar songs, we care about direction, not intensity. We will develop a solution to this problem after you answer some questions:

The Spotify dataset we are using represents songs by a vector $v \in \mathbb{R}^9$. Assuming each component in $v$ is independent, what directions does this nine-dimensional vector represent and how would this change if the features in the vector are highly correlated?



**If each component in $v$ is independent, the direction represents each track's unique combination of attributes. For instance, a vector that's heavy in Danceability and Energy might point to an EDM artist like Martin Garrix.**



**If the features in the vector are highly correlated, then we would expect something like Loudness to scale with Energy. This means there are actually less than 9 unique attributes that give a song its overall style. Therefore, we would expect song vectors to cluster together in distinct regions $\mathbb{R}^9$ that correspond to different genres. For example, a song vector pointing in the Danceability direction might correspond to high Energy and low Acousticness, which creates a region corresponding to EDM music.**

## Normalization

To make Song A and Song B comparable from our example above, we need to remove the influence of magnitude. We do this by normalizing every vector in our dataset to be
$$ \mathbf{u} = \frac{\mathbf{v}}{\| \mathbf{v} \|} $$

Every song vector is shrunk or stretched until it has a length of exactly 1. By forcing all songs to live on the same "surface", distance is now purely a measure of the angle between styles.

Before normalization:
1. $\textbf{Song A (Loud):}$ $\mathbf{v}_A = \begin{bmatrix} 10 \\ 4 \end{bmatrix}$ (High Volume, Moderate Tempo)
2. $\textbf{Song B (Quiet):}$ $\mathbf{v}_B = \begin{bmatrix} 5 \\ 2 \end{bmatrix}$ (Med Volume, Moderate Tempo)

After normalization, the vectors are identical:

$$\mathbf{v}_{A/B} = \begin{bmatrix} 0.928 \\ 0.371 \end{bmatrix}$$

*Note that we first standardized the song vectors to get each of the nine features on the same scale. We will not cover this in any detail.

## Ready for Dot Products

Now that every song is a unit vector, we have unlocked a special property of the dot product. Section 1.2 of Strang defines the geometric definition of the dot product as

$$
\mathbf{v}\cdot\mathbf{w}=\lVert \mathbf{v} \rVert\,\lVert \mathbf{w} \rVert \cos\theta,
$$

which reveals the utility of the operation. If vectors are normalized such that $\lVert \mathbf{v} \rVert = 1$, then the dot product simplifies to $\cos\theta$. This means the dot product of two song vectors gives us the Cosine Similarity directly.

Your first Python task is to compute the vector norms (normalize) the features in the Spotify dataset.


```python
# Implement normalization to prevent "volume bias" where loud tracks dominate
# search results regardless of style.

# Compute row-wise L2 norms
norms = np.linalg.norm(A, axis=1, keepdims=True)
A_normalized = A/norms
# np.linalg.norm computes Euclid. norm (length), axis=1 computes norm across
# each row (not column), and keepdims=True keeps shape (m,1) instead of (m,).
# A/norms normalizes each row vector by dividing by its length.

# Validation: Uncomment to verify unit length
check_norms = np.linalg.norm(A_normalized, axis=1)
assert np.allclose(check_norms, 1.0), "Normalization failed."

# Query: uncomment the next two lines and replace with your fav artist
query_idx = df.index[df['track_artist'] == 'Coldplay']
v_query = A_normalized[query_idx, :]

# Similarities to each query track
S = A_normalized @ v_query.T

# Aggregate to a single score per track
scores = S.mean(axis=1)

df['similarity'] = scores

top_10 = (df
          .sort_values('similarity', ascending=False)
          .loc[:, ['track_name', 'track_artist', 'similarity']]
          .head(10))

bottom_10 = (df
             .sort_values('similarity', ascending=True)
             .loc[:, ['track_name', 'track_artist', 'similarity']]
             .head(10))

print("=== Top 10 ===")
print(top_10.to_string(index=False))

print("\n=== Bottom 10 ===")
print(bottom_10.to_string(index=False))
```

### Feature Isolation

So far, we have relied on the nine features provided in the Spotify dataset. We can easily answer which song is the most acoustic by looking at that component in our vector. What if we want to ask more complex questions like "what are the vocals like is this song?" There is no single column for vocals. To do this, we construct a new axis in our 9D space which points towards the abstract concept of "vocals".

In Section 4.2, Strang will introduce the projection matrix

$$
p=\frac{aa^{\mathsf T}}{a^{\mathsf T}a}
$$

which projects a vector onto the line spanned by $a$. In the context of music search, this allows for the isolation of specific components of the song. The 'vocal axis' in the vector space we will create is represented by
the vector $\mathbf{v}_{\text{vocal}}$.

Imagine you have a song vector $\mathbf{b}$ that contains both vocal and instrumental components. The vocal component can be isolated by projecting $\mathbf{b}$ onto the vocal axis. We want to break the vector $\mathbf{b}$ into two perpendicular components:
$$ \mathbf{b} = \mathbf{p} + \mathbf{e} $$

1. $\mathbf{p}$ is the Vocal Component (the part of the song parallel to our vocal archetype).
2. $\mathbf{e}$ is the Instrumental Component (the part of the song orthogonal to the vocals).

### Calculating the Projection

The projection of a vector $\mathbf{b}$ onto a unit vector $\mathbf{v}$ is the "shadow" that $\mathbf{b}$ casts along that direction. Since $\mathbf{v}_{vocal}$ is a unit vector, the formula for the projection $\mathbf{p}$ is:

$$ \mathbf{p} = \text{proj}_{\mathbf{v}_{vocal}}\mathbf{b} = (\mathbf{b} \cdot \mathbf{v}_{vocal}) \mathbf{v}_{vocal} $$

Recall from previously that the dot product $(\mathbf{b} \cdot \mathbf{v}_{vocal})$ is a scalar that tells us the similarity between the song and the vocal archetype. If the dot product is high, the scalar is large, and the shadow $\mathbf{p}$ is long meaning the song is very vocal. If the dot product is zero, the scalar is zero, and $\mathbf{p} = \mathbf{0}$.

Once we have the vocal part $\mathbf{p}$, finding the instrumental part is a simple matter of vector subtraction. This is the error vector:
$$ \mathbf{e} = \mathbf{b} - \mathbf{p} $$

Geometrically, we have subtracted the "vocal-ness" out of the song.

Note: In class, we learned that the null space of a transformation is the set of all vectors that map to zero. Any song that is already purely instrumental will have a projection $\mathbf{p} = \mathbf{0}$.  Therefore, purely instrumental songs live in the null space of the projection onto $\mathbf{v}_{vocal}$.

Your task is to read the code below, then implement the projection of $b$ onto $a$, the vocal component. The most important property of this decomposition is that p and e must be orthogonal. If your
math is correct, the dot product of these two new vectors must be zero. The rest of the code is already complete.



```python
# Step 1: Define the Vocal Axis (a)
# We define 'Vocalness' axis using tracks with high speechiness (> 0.66)
mask_vocal = df["speechiness"].to_numpy() > 0.66
vocal_samples = A_normalized[mask_vocal]
v_vocal = np.mean(vocal_samples, axis=0)
v_vocal = v_vocal / np.linalg.norm(v_vocal)

# Step 2: Select a target song (b) to instrumentalize
# Picking a track with high vocal presence
b = A_normalized[139, :]

# Step 3: Project b onto a (The 'Vocal' component)
# Formula: p = a * (a^T * b / a^T * a)
# YOUR CODE HERE

# Step 4: Isolate the Instrumental Error Vector (e)
# Formula: e = b - p (The part of b orthogonal to the vocal axis)
e = b - p

# Step 5: Verification of Orthogonality
ortho_check = np.dot(e, v_vocal)
print(f"Orthogonality Check (e dot a): {ortho_check:.10f}")
```

### Visualization and Dimensionality Reduction

Finally, now that we have embedded our songs and artists in $\mathbb{R}^{9}$ how can we visualize our embeddings of songs? Humans are limited to three spatial dimensions, yet AI embeddings often operate in hundreds or thousands. Principal Component Analysis (PCA) is the primary tool used by AI engineers to 'squash' these high-dimensional spaces into a form that can be visually inspected while retaining the most important patterns in the data. We will learn about PCA in future sections of the course. Run the code below for your favorite set of artists (minimum 10) and interpret the results.

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

artists_10 = ['Coldplay', 'Drake', 'Ed Sheeran', 'Taylor Swift', 'The Weeknd',
              'Billie Eilish', 'Post Malone', 'Ariana Grande', 'Kendrick Lamar',
              'Maroon 5', 'Imagine Dragons']

mask = df['track_artist'].isin(artists_10).to_numpy()
labels = df.loc[mask, 'track_artist'].to_numpy()

def plot_cloud_and_centroids(Z, title):
    plt.figure(figsize=(10,7))
    plt.scatter(Z[:, 0], Z[:, 1], s=10, alpha=0.10)

    # centroids
    for artist in artists_10:
        pts = Z[labels == artist]
        if len(pts) == 0:
            continue
        c = pts.mean(axis=0)
        plt.scatter(c[0], c[1], s=250, edgecolors='black')
        plt.text(c[0], c[1], " " + artist, fontsize=9)

    plt.title(title)
    plt.grid(alpha=0.2)

    # Fixed axis limits
    plt.xlim(-4, 3)
    plt.ylim(-3, 3)

    plt.tight_layout()
    plt.xlabel("Component 1 (Acoustic -> Produced)")
    plt.ylabel("Component 2 (Soft -> Energetic)")
    plt.show()

# PCA on standardized features only
Z_scaled = PCA(n_components=2, random_state=0).fit_transform(A[mask])

plot_cloud_and_centroids(Z_scaled, "Principal Component Analysis")
```

**The PCA plot compresses the 9D song space into 2 dimensions while it preserves most of the variance.
